# Custom Tools


# Method1 :Using Tool decorator


```python
from langchain.tools import tool
```

```python
# Step 1  Creating a function

def multiply(a , b):
    """Multiply two numbers """
    return a*b
```

```python
#Step 2 : add type hints 
def multiply(a : int, b: int )->int :
    """Multiply two numbers"""
    return a*b
```

```python
# Step 3 : add tool decorator
@tool
def multiply(a : int, b: int )->int :
    """Multiply two numbers"""
    return a*b
```

```python
result=multiply.invoke({'a':6 ,'b':7 })
result
```

```python
print(multiply.name)
print(multiply.args)
print(multiply.description)
```

```python
print(multiply.args_schema.model_json_schema())
```

# Method 2 : Structured Tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel , Field
```

```python
class MultiplyInput(BaseModel):
    a: int =Field(required=True ,  description= "First Number" )
    b: int = Field(required=True, description='Second number')
    
```

```python
def multiply_func(a : int , b : int )-> int:
    return a*b
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply_func,
    description='To Multiply two numbers ',
    name='multiply',
    args_schema=MultiplyInput
)
```

```python
result=multiply_tool.invoke({'a':5,'b':9})
print(result)
print(multiply_tool.name)
```

# Method 3 : Using BaseTool Class

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool=MultiplyTool()
result=multiply_tool.invoke({'a':4 , 'b':235})
result
```

# ToolKit


```python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolKit:
    def get_tools(self):
        return [add , multiply]
        
```

```python
toolkit = MathToolKit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
