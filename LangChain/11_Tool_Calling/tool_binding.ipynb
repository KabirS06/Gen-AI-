```python
from langchain_huggingface import ChatHuggingFace , HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
```

```python
load_dotenv()
```

```python
# Creating tool 
@tool
def multiply(a :int , b:int)->int:
    """When given 2numbers return their product"""
    return a*b
```

## Tool Binding
* Only limited LLMs which can be binded with tools 

```python
llm=HuggingFaceEndpoint(model='meta-llama/Llama-3.1-8B-Instruct',
                        task='text_generation')
model=ChatHuggingFace(llm=llm)
```

```python
llm_with_tool=model.bind_tools([multiply])
```

```python
llm_with_tool.invoke('h')
```
