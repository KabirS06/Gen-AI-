```python
from langchain_huggingface import ChatHuggingFace , HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
```

```python
load_dotenv()
```

```python
# Creating tool 
@tool
def multiply(a :int , b:int)->int:
    """When given 2numbers return their product"""
    return a*b
```

## Tool Binding
* Only limited LLMs which can be binded with tools 

```python
llm=HuggingFaceEndpoint(model='Qwen/Qwen3-32B',
                        task='text_generation')
model=ChatHuggingFace(llm=llm)
```

```python
llm_with_tool=model.bind_tools([multiply])
```

```python
llm_with_tool.invoke('Hi how are you ?')
```

```python
query=HumanMessage("Multiply 6 with 7")
```

```python
message=[query]
message
```

```python
result=llm_with_tool.invoke(message)
result
```

```python
message.append(result)
```
